##### 1.5 Measue the shapes
1. Skewness = Which direction is the data stretched? (data kis side jhuk raha hai?)
2. Kurtosis = Where is particular percentage of data? (data mein hum kis position par hain?)
3. Quantiles = How heavy are the tails / how common are extreme values? (data ke tails/extreme values kitne heavy hain?)


* Skewness:-
1. if Mean < Median < Mode, then Skew < 0. It means Negative skewed (Left Skew)
2. if Mean > Median > Mode, then skew > 0. It means Positive Skewed (Right Skew)
3. if Mean = Median = Mode, then Skew = 0. It means Data distribution is Symmetric.

```python
# Skewness of Outcome column
# skewness = df['Outcome'].skew()
#print(f"Skewness of Outcome column: {skewness}")
print(f"Median: {df['Outcome'].median()}, Mean: {df['Outcome'].mean()}, Mode: {df['Outcome'].mode()[0]}")
# Outcome Column Mean>Median>Mode, so the distribution is positively and right skewed. The tail on the right side of the distribution is longer or fatter than the left side.
print(f"Median: {df['Glucose'].median()}, Mean: {df['Glucose'].mean()}, Mode: {df['Glucose'].mode()[0]}")
# Glucose Column Mean>Median>Mode, so the distribution is positively and right skewed. The tail on the right side of the distribution is longer or fatter than the left side.
print(f"Skewness of Glucose column: {df['Glucose'].skew()}") # Skewness of Glucose column: 0.1737547243077745 which is greater than 0
```
