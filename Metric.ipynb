# In Machine Learning (ML), metrics are the measures we use to evaluate how well a model is performing. They act like a "scorecard" for your model, telling you whether its predictions are accurate, useful, and aligned with the problem type.

## Regression Metrics:-
1. MAE (Mean Absolute Error):- It is an absolute difference error between actual and predicted values. It's unit is same as target variable
2. MSE (Mean Squared Error):- It is an average of squared difference of errors between actual and predicted values.. It's unit is in squared of target variable
3. RMSE (Root Mean Squared Error):- 
4. R2 SCORE:- R-Squared 
5. ADJUSTED R2 SCORE

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score

print("All Important Library is imported Successfully!")
```

```python
df = pd.read_csv('placement.csv')
df.head(10)
```

```python
X = df.drop(columns='package')
y = df['package']

X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=2,test_size=0.2)

lr = LinearRegression()

lr.fit(X_train,y_train)

pred = lr.predict(X_test)

print("MAE:",mean_absolute_error(pred,y_test))
print("MSE:",mean_squared_error(pred,y_test))
print("RMSE:",np.sqrt(mean_squared_error(pred,y_test)))
print("R2 Score:",r2_score(pred,y_test))
r2 = r2_score(pred,y_test)

r2_adj = 1-((1-r2)*(40-1)/(40-1-1))

print("Adjusted R2:", r2_adj)
```

### For Adjusted R2 score we have formula:-
#### 1 - ((1-r2)*(n-1)/(n-k-1))
* Where:-
1. r2 = r2_score
2. n = total data points 
3. k = number of independent variables
